```python
import pandas as pd
```

```python
df = pd.read_csv("6.1_raw_data.csv")
```

## Feature Engineering
- ### It means tranforming the data to create some new column or new feature that improve the performance of the model.

```python
# 1. apply()
# apply() function which take function as parameter
# this function is called high order function it means
# it take function as parameter and can return the function
# example : add a new column for tax 20% or income > 50k and 10% below that.
df2 = df.copy()

# hum df2 ke under tax columns create karege = df2 ke income ke uper ek function apply karenge 
df2["tax"] = df2["income"].apply(lambda x : "20%" if x > 50000 else "10%")
df2
```

```python
# 2. map()
# map() -> existing value ko map karna cahate hai 
gender_map = {"Male":"M", "Female" : "F", "Unknown": "U"} 
df2["gender"] = df2["gender"].map(gender_map)
df2
```

```python
# 3. assign() function -> important in sytax
# assign a new column
# create a new_income and having 10% increment
df2 = df2.assign(new_income = df2["income"] * 1.1)
df2
```

```python
 # 4. replace(old,new)
df2["country"] = df2["country"].replace("USA","United State")
df2
```

```python
# rename Columns & row
df2.columns
# Capitalize the columns name
df2.columns = ["Id","Name","Age","Country","Gender","Income", "Tax","New_Income"]
df2.rename(columns={"Income":"Salary"})
df2.rename(index = {1 : "First"})


# Sorting
df2.sort_values("Income") #asending order 
df2.sort_values(["Income"], ascending=False) #descending order 
# if income se then sort using age
df2 = df2.fillna(50)
sorted_df = df2.sort_values(["Income","Age"])  # se the (10,8) label

sorted_df.sort_index()

# Reset Index
sorted_df.reset_index()  # preserved the original index
sorted_df.reset_index(drop=True) # remove old index

# Ranking
sorted_df["Ranking"] = sorted_df["Income"].rank()
# tie hogaye gi toh average aayega
# do we want dense number means integer
sorted_df["Ranking"] = sorted_df["Income"].rank(method="dense")
# sorted_df["Ranking"] = sorted_df["Income"].rank(method="min")
# sorted_df["Ranking"] = sorted_df["Income"].rank(method="max")
sorted_df


# rearrange the columns
# tax column will come on last
sorted_df[["Id", "Name" , "Age", "Country" , "Gender" , "Income", "New_Income", "Ranking", "Tax"]]
```
