```python
import pandas as pd
```

```python
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))

# Indexing access 
print(s[0])
print(s[3])
```

```python
s = pd.Series([23,24,25,26], index=["Adam", "piyush", "Eve", "Om"])
print(s)

# value access
print(s["piyush"])
# index access -> give warning (using iloc to access using index)
print(s[1])
print(s.iloc[1])
print(s.index)
```

```python
# Properties
# 1. Homogenous
s = pd.Series([1,2,3,4,"piyush", 3.14]) # dtype = object
print(s)
```

```python
# 2. Vectorization
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])
print(s1 + s2)
```

```python
# 3. Handle Missing Value with NaN => None
```

```python
# 4. mutable values, immutable size
s = pd.Series([10,20,30,40,50])
s[0] = 100
print(s)

# immutable size => if we add or delete element then new series will be created
new_series = s.drop(0)
print(new_series)
print(s)
```
