import pandas as pd

```python
df = pd.read_csv("Titanic-Dataset.csv")
df.head(3)
```

### A. Display only columns: Name, Sex, Age, Fare, Survived

```python
df.columns
```

### B. Select passengers who:
- Are female
- Fare > 30

```python
selected_df = df[(df["Sex"].str.lower() == "female") & (df["Fare"] > 30)]
selected_df.head()
```

### C. Group by Pclass and compute:
- Survival rate
- Average Fare
- Average Age

```python
df.groupby("Pclass").agg(
    survival_rate = df["Survived"].count()
)
```
