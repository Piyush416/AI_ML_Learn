```python
import pandas as pd
```

```python
df_customers = pd.DataFrame({
    "customer_id" : [1,2,3,4],
    "name" : ["Adam" , "Bob" , "Charlie", "Dave"],
})

df_orders = pd.DataFrame({
    "order_id" : [101,102,103,104],
    "customer_id": [2,1,4,5],
    "amount" : [250,120,300,180]
})

```

```python
# Mergering and Joining

#inner join 
# unqiue column se match kar ke common data rakh lenge
pd.merge(df_customers, df_orders, on="customer_id")

# left join
# issme left data toh rahega or uuske corresponding NAN honga 
pd.merge(df_customers, df_orders, on="customer_id", how="left")

# right join
# iisme right data toh rahega or uuske corresponding NAN honga
pd.merge(df_customers, df_orders, on="customer_id", how="right")

# outer join
# both data rahega left and right 
pd.merge(df_customers, df_orders, on="customer_id", how="outer")
```
