```python
import pandas as pd
df = pd.read_csv("6.1_raw_data.csv")
```

```python
# Data Cleaning
# 1.handle missing values
df.isnull()  # show true if NaN present(Not a Number)
df.isnull().sum()


df.dropna() #give only those row which have no NaN
# df.dropna(axis = 1) #give columns

df.fillna(0)
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data

df.ffill() #forward fill top value niche copy ho jayegi NAN me 
df.bfill() #backward fill from bottom to top fill
```

```python
# 2. Handle Duplicates

df.duplicated()
df["country"].duplicated()
df[["country","gender"]].duplicated()

df.drop_duplicates()
df.drop_duplicates(inplace = True) #original dataset me change ho jayega
```

```python
# 3. Handle Datatypes & Datetime

# change datatypes using astype function
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64").copy()
df2.dtypes

# to_datetime() function
date_str = pd.Series(pd.to_datetime("26-12-30"))
type(date_str.dtypes)
```

```python
# 4. Handle String
df["gender"].str.capitalize()
df["gender"].str.upper()
df["name"].str.split(" ")
df["country"].str.contains('india', case=False)
```
