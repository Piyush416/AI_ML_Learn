```python
import pandas as pd
```

```python
# DataFrame
# Creation
# 1. dictionary of list
info = {
    "Name" : ["piyush", "Shree", "Om"],
    "CGPA" : [9.3, 9.8, 7.6],
    "Age" : [22,21,26]
}

df = pd.DataFrame(info)
print(df, type(df))
print(df.index)
print(df.columns)
```

```python
# 2. list of list
import numpy as np
np_arr = np.array([["adam",23],["bob",25],["piyush", 24]])
df = pd.DataFrame(np_arr, columns = ["Name","Age"])
df
```

```python
# Read JSON
df = pd.read_json("4_employee_data.json")
print(df, type(df))
```

```python
# Read CSV
df = pd.read_csv("4_employee_data.csv")
print(df, type(df))
```
