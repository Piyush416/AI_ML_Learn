```python
import pandas as pd
```

```python
df = pd.read_csv("6.1_raw_data.csv")
```

```python
# remove duplicated , fillna  with 0, and sort by income
# and store this into new csv file
df2 = df.copy()
df2 = df2.drop_duplicates()
df2 = df2.fillna(0)
df2 = df2.sort_values("income")
df2 = df2.reset_index(drop=True)
df2

# write this data to csv file
df2.to_csv("8.1_sorted_data.csv")
```
