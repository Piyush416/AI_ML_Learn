```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "country" : ["USA","USA","INDIA","INDIA"],
    "year" : [2020,2021,2020,2021],
    "sales" : [100,120,90,110],
    "profile": [20,25,18,22]
})
df
```

```python
# Melt function use to reshaping the data
# Melt help to convert wide data to long formate (means horitonal to vertical)
# syntax melt(id_vars, value_vars, var_name, var_value)
melted_df = df.melt(
    id_vars = ["country", "year"],  #ye column change nhi honge
    value_vars = ["sales", "profile"], #ye column ko melt karna hai 
    var_name = "metric",  #new_column name
    value_name = "value"   #new_column value name
)
```

```python
# Pivot use for reverse of melt long formate to wide 
melted_df.pivot(
    index = ["country", "year"], #index must be unqiue value so we use country and year pair to make new 
    columns= "metric", #hum metric se column create karege 
    values= "value",  #her ek column ki value value denge
)
```
