```python
import matplotlib.pyplot as plt
```

```python
X = [1,2,3,4]
Y = [5,6,7,8]

plt.plot(X,Y) #Line Plot
```

```python
# Real Data

#Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
year = [2008,2009,2010,2011,2012]
oscar_revenue = [1005, 170, 427, 133, 232]

#plot
plt.plot(year, oscar_revenue)
plt.title("Oscar Movie Revenue Each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non Oscar Winner Movie
non_oscar_movies = ["slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(year, oscar_revenue, "o--r",label="Oscar Movie")  #"o--r" is format string -- line use , red color
plt.plot(year, non_oscar_revenue, color="blue", marker=">", linestyle="-.",label="Non Oscar Movie") # we can also write seperate formate string
plt.title("Oscar Movie vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend() # this will show the label of plot (a block in top left side)
# plt.grid()
```

```python
# Movies and Revenue
plt.plot(oscar_movies, year) # now you see the movie name is mixing. 
plt.tight_layout() # using this we can fix the overlapping names.
plt.savefig("final_plot.png") # to save the figure 
```
