```python
import matplotlib.pyplot as plt
```

```python
people = ["Person A","Person B","Person C","Person D","Person E","Person F","Person G","Person H","Person I","Person J"]
age = [22,25,30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150]

colors = ["red" if x > 135 else "green" for x in bp]

# scatter plot is use to 
# 1. correlation
# 2. outliner (koi ek point range se alag hai)
# 3. cluster (multi datasets me pata chalega)

plt.scatter(age, bp, color = colors, s=bp, marker = "<", alpha=0.2)  # s=size #alpha => opacity
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()

```

```python
# colormap in scatter plot => range of color
plt.scatter(age, bp, cmap="OrRd", c=bp)  #OrRd -> orangeRed
plt.colorbar(label = "BP")
plt.grid()

# add annotate
# plt.annotate("person C", xy = (30,120), xytext=(30+2,120))

# i want annotate at each point
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i], bp[i]+2))
    
plt.ylim(min(bp), max(bp)+5)
plt.xlim(min(age), max(age)+10)
```

```python
# Multiple Plots

cities = ["City A", "City B", "City C", "City D", "City E"]

#Winter: Temperature (°C) vs Humidity(%) # ° (option + shift + 8) keyboard shortcut
winter_temp = [5,2,10,0,7]
winter_humidity = [80,75,65,85,70]

#Summer: Temperature(C) vs Humidity(%)
summer_temp = [25,30,28,35,27]
summer_humidity = [60,50,55,45,65]

# this is use to see the cluster 
plt.scatter(winter_temp, winter_humidity, label="Winter")
plt.scatter(summer_temp, summer_humidity, label="Summer")

plt.title("Summer vs Winter Data")
plt.ylabel("Humidity(%)")
plt.xlabel("Temp(°C)")
plt.legend(loc="lower left")
```
