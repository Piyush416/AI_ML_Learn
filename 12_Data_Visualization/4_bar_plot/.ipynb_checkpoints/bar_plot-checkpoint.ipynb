```python
import matplotlib.pyplot as plt
```

```python
# Pie Charts -> relations batana ho iphone vs andriod user 

expenses = ["Salaries","Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500,150,120,100,50]

plt.pie(amounts, labels=expenses)
# plt.style.available  #shows all the style name
plt.style.use("dark_background")
```

```python
plt.style.use("default")

expenses = ["Salaries","Rent", "Marketing", "R&D", "Miscellaneous"]
explodes = [0,0,0,0.2,0]

plt.pie(amounts, labels=expenses, autopct="%1.1f%%",
       wedgeprops = {
           "edgecolor": "black",
           "linewidth": 1,
           "linestyle": "--",
           # "alpha" : 0.2
       },
        explode = explodes,
        startangle  = 360,
        shadow = True
       )  # autopercentage -> autopct
```
