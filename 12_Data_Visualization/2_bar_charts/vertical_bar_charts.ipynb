```python
import matplotlib.pyplot as plt
```

```python
#Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
year = [2008,2009,2010,2011,2012]
oscar_revenue = [1005, 170, 427, 133, 232]

# vertical Bar X-axis => categories or Y-axis => values

plt.bar(year, oscar_revenue)
plt.title("Revenue in Each Year")
plt.xlabel("years")
plt.ylabel("Revenue in ($M)")

# now i want to give the movie name on this bar
# plt.text(year[0], oscar_revenue[0], "ABC", ha="center") # position x, position y, "text" , alignment

# using for loop 
for i in range(len(year)):
    plt.text(year[i], oscar_revenue[i]+10, oscar_movies[i], ha="center") # y axis me thoda uper lene ke liye y+10 karege


# increase the graph size or graph limit
plt.ylim(0, max(oscar_revenue)+100)
```

```python
# Compare Two DataSets
#Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
year = [2008,2009,2010,2011,2012]
oscar_revenue = [1005, 170, 427, 133, 232]

# Non Oscar Winner Movie
non_oscar_movies = ["slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

import numpy as np
x = np.arange(len(year)) # create a array from 0 to size of year ([0,1,2,3,4])

# by default width of each bar 0.8 hoti hai or center 0.5 hota hai
width = 0.4 # toh humne width ko half kar diya or x se + or - kar diya 

plt.bar(x+width/2, oscar_revenue, width, label="Oscar Movies")
plt.bar(x-width/2, non_oscar_revenue,width, label="Non Oscar Movies")
plt.legend()

# now in x axis we want years so we use 
plt.xticks(x,year)
```
