```python
import json
```

```python
# Load the data 
def load_data(filename):
    with open(filename, 'r') as f:
        store_data = json.load(f);

    return store_data;
    
```

```python
data = load_data("store_data.json")
```

```python
print(data)
print(type(data))
```

```python
# Clean and structure the data

def clean_data(data):

    text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5}
    cleaned_data = []
    unique_user = set()
    
    for user in data:
        # clean rating -> data consistency
        raw_rating = str(user["rating"]).strip().lower()  #strip means clean the with space from back and front
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        # Handle missing vals
        raw_age = user.get("age")  # we use get method it will give None not error if not present
        if(raw_age == None):
            user["age"] = None

        # Deduplication
        if(user['name'].strip() in unique_user):
            continue;
            
        unique_user.add(user['name'])
        cleaned_data.append(user)

    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

```python
# get meaningful insights from data
def get_insights(data):

    # avg rating
    total_rating = 0
    for user in data:
          total_rating += float(user['rating'])
    print(f"avg rating: {total_rating/len(data)}")

    # percentage of user with poor rating
    poor_rating = 0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_rating+=1
    print(f"% of user with poor rating: {poor_rating/len(data) * 100}%")
```

```python
get_insights(data)
```

```python
# Build Recommendation Feature
def get_recommendations(data):
    recommendations = []
    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "apple"
        else:
            curr_recomm["brand"] = "samsung"

        recommendations.append(curr_recomm)
    return recommendations
```

```python
get_recommendations(data)
```
