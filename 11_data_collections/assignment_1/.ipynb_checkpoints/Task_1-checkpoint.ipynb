```python
import requests
from bs4 import BeautifulSoup
```

```python
page_count = 1001

while True:
    res = requests.get(f"https://quotes.toscrape.com/page/{page_count}/")
    soup = BeautifulSoup(res.text, 'lxml')
    quotes = soup.select('div.quote')
    if not quotes:
        print("No valid pages anymore..")
        break

    with open(f"scraped_data/quotes{page_count}.html", "w") as f:
        f.write(res.text)
        print(f"downloaded data from page{page_count}")
        
    page_count = page_count+1
        
```

```python
# Extract useful info 
all_life_quotes = []
for i in range(1,11):
    with open(f"scraped_data/quotes{i}.html", 'r') as f:
        html_content = f.read()

    # parse the html code using bs4
    soup = BeautifulSoup(html_content, 'lxml')

    all_quotes = soup.select('div.quote')
    life_quotes = []
    for q in all_quotes:
        all_tags = []
        for tag in q.select('.tags .tag'):
            all_tags.append(tag.get_text())
        if "life" in all_tags:
            text = q.select_one('span.text').get_text()
            autor = q.select_one('small.author').get_text()
            life_quotes.append([text,autor])
    all_life_quotes.append(life_quotes)
# print(all_life_quotes)
   
```

```python
import pandas as pd

flat = []
for sublist in all_life_quotes:
    for item in sublist:
        flat.append(item)

df = pd.DataFrame(flat, columns = ["quote", "author"])
df.to_csv("cleaned_data/data.csv", index=False)
```
