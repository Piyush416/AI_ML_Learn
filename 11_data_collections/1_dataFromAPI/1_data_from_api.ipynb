```python
import requests  # using this we request the api
```

```python
URL = "https://stephen-king-api.onrender.com/api/books" #this is api
```

```python
res = requests.get(URL)
res  # it will print the status of api
```

```python
json_data = res.json()
# json_data // printing the json data 
```

```python
# convert the json data to data frame
import pandas as pd
df = pd.json_normalize(json_data["data"])  #json_data ke under ek key hai data use me sara data store hai 
df = df[['id','Title','Year','ISBN']]  #fitler the data 
```
