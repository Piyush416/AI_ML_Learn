```python
from bs4 import BeautifulSoup
```

```python
with open('scraped_data/data1.html', 'r') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'lxml')  # here we create a object using BeautifulSoup class constructor and pass html file to it 
```

```python
# soup.find("h3") #give single html tag
all_h3 = soup.find_all("h3") #give all h3 tag in array 

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True) # give the text inside h3 tag
    # print(h3.find_parent("div")["class"])

    # to get next sibling of h3
    population = h3.find_next("div").select('span.country-population')[0].get_text(strip=True)
    # print(h3.find_next("div").select_one('span.country-population').get_text(strip=True)) 

    # to get all attributes
    # print(h3.attrs)
    all_countries.append([name, population])
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns = ["Name","Population"])
```

```python
df.to_csv("cleaned_data/data.csv", index=False)
```
