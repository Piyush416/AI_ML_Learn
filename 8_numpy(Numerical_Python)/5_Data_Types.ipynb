```python
import numpy as np
```

```python
# Integer datatypes => int8, int16, int32, int64, int128
arr = np.array([1,2,3,4,5])
print(arr, arr.dtype)
```

```python
# floating -> float64, float128
arr1 = np.array([1.1, 2.3])
print(arr1, arr1.dtype)
```

```python
# complex nums. -> complex64, complex128
arrc1 = np.array([6 + 4j])
arrc2 = np.array([3 + 2j])
print(arrc1 + arrc2 , arrc1.dtype)
print(arrc1 - arrc2 , arrc1.dtype)
```

```python
# String: S(byte-str) & U(unicode-str)
arrs1 = np.array(["piyush", "shree"])
print(arrs1 , arrs1.dtype)
```

```python
# Object -> Heterogenous converted into object
arrobj = np.array([1, "prime", {1,2,3,4}])
print(arrobj, arrobj.dtype)
```
