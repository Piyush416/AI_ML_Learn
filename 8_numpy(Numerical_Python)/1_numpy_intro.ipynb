```python
# Why numpy is faster ?
# 1. Homogenous Data types and contiguous memory
# 2. vector Operations and C implementations
# 3. Reduced Overhead (python jese meta data store nahi karta hai)

import numpy as np
import time
```

```python
# execution performance
size = 10_000_000  #largest data set of 10million numbers

#python list
py_list = list(range(size))
start = time.time();
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"python list time = {end-start} seconds")

# numpy arrays
np_arr = np.array(py_list)
start = time.time()
sq_array = np_arr ** 2  #vectorization
end = time.time()
print(f"np array time = {end-start} seconds")
```

```python
#memory 
import sys
print(f"python list size = { sys.getsizeof(py_list) * len(py_list) } bytes")
print(f"numpy array size = {np_arr.nbytes} bytes")

# why this largest gap happened because python store data in non contiguos block in memory. 
# and it will store heterogenous data.
# so it have to store metadata of list -> next address and datatypes and all.

# np_array is homogenous data store and it will store data in contiguos block in memory.
```
