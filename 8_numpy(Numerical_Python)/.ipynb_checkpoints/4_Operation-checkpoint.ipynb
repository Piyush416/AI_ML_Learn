```python
import numpy as np
```

```python
# Operations in Array

# Reshaping
arr = np.array([[1,2,3],[4,5,6]])
print(arr, arr.shape)

reshaped = arr.reshape((3,2))
print(reshaped, reshaped.shape)

# Flatten
flatten = arr.flatten() # 2D => 1D
print(flatten, flatten.shape)
```

```python
# Fancy Indexing  -> Access multiple index at onces
arr = np.array([1,2,3,4,5])
inx =  [0,2,3]
print(arr[inx])

# Boolean Indexing -> condition in indexing
print(arr[arr > 4])
print(arr[arr % 2 == 0])
print(arr[arr % 2 != 0])
```
