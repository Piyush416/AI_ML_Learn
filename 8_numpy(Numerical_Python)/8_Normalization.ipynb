# Normalization
## Normalizing an array means transforming its values so they fit into a specific scale or range
## use in machine learning model to improve there performance

## standard Normalization -> very popular
## formula
- (arr-mean)/std_dev
- ## std_dev => standard Deviation
- formula std_dev
- sqrt(sum((value(i) - mean)^2 / N))

```python
import numpy as np
```

```python
# Normalization
arr = np.array([[1,2],[3,4]])
mean = np.mean(arr)
print(mean)
std_dev = np.std(arr)
print(std_dev)
normalized_arr = (arr-mean)/std_dev
print(normalized_arr)
```
